```python
import pandas as pd
```

```python
df = pd.read_csv('Heart.csv')
```

```python
resting_bp_median = df[df['RestingBP'] != 0]['RestingBP'].median()
```

 RestingBP: replace the 0 with the median of all valid (non-zero) values


```python
df['RestingBP'] = df['RestingBP'].replace(0, resting_bp_median)
```

```python
cholesterol_median = df[df['Cholesterol'] != 0]['Cholesterol'].median()
df['Cholesterol'] = df['Cholesterol'].replace(0, cholesterol_median)
```

Cholesterol: same approach
