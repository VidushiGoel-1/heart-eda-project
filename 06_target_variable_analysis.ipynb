```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('Heart.csv')
```

 Count how many patients fall into each HeartDisease category


```python
df['HeartDisease'].value_counts()
```

 Same thing, but as percentages instead of raw counts


```python
df['HeartDisease'].value_counts(normalize=True) * 100
```

```python
sns.countplot(x=df['HeartDisease'])
plt.title('HeartDisease Class Balance')
plt.show()
```

if one class massively outnumbers the other (like 95% vs 5%), it's called an "imbalanced dataset" & it becomes a problem later when building a model, since the model can just predict the majority class every time and still look accurate. Checking this now, during EDA, is how you catch that risk early.
