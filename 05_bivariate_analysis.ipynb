Block 1: imports and repeate 

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('Heart.csv')
df['RestingBP'] = df['RestingBP'].replace(0, df[df['RestingBP'] != 0]['RestingBP'].median())
df['Cholesterol'] = df['Cholesterol'].replace(0, df[df['Cholesterol'] != 0]['Cholesterol'].median())
```

Does Age relate to HeartDisease?


```python
sns.boxplot(x=df['HeartDisease'], y=df['Age'])
plt.title('Age by HeartDisease status')
plt.show()
```

 Correlation heatmap — shows how strongly numeric columns relate to each other


```python
numeric_df = df.select_dtypes(include='number')
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

Oldpeak   : this one should show a positive correlation with HeartDisease (a warm/red cell). Oldpeak measures ST depression from an exercise stress test — a higher value generally indicates more cardiac stress. So: higher Oldpeak → more associated with having heart disease.

MaxHR     : this one should show a negative correlation (a cool/blue cell). MaxHR is the maximum heart rate achieved during exercise. So: lower MaxHR → more associated with having heart disease — makes sense too, since a heart under strain often can't reach as high a max rate during exertion
